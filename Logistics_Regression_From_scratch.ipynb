```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification

x_raw , y_raw = make_classification(n_samples = 3000,n_features=5,n_redundant=0,n_classes=2,random_state=42,n_clusters_per_class=1)

```

```python
y = y_raw.reshape(-1,1)
print(x_raw.shape)
print(y.shape)

learning_rate = 0.1
iterations = 3000
```

```python
def sigmoid_call(W,X,B):
  return 1/(1+np.exp(-((X @ W) +B)))
```

```python
features = x_raw.shape[1]

in_w = np.zeros((features, 1))
in_b = 0

print(in_w.shape)
```

```python
for i in range(iterations):
  sig = sigmoid_call(in_w,x_raw,in_b)
  loss = -(y*(np.log(sig)) + (1-y)*(np.log(1-sig)))
  cost = np.mean(loss)
  dw = (x_raw.T @ (sig - y))/(x_raw.shape[0])
  db = np.sum((sig - y))/(x_raw.shape[0])
  in_w = in_w - learning_rate*(dw)
  in_b = in_b - learning_rate*(db)
```

```python
print(in_w)
```

```python
print(in_b)
```

```python
def predict(X, w, b, threshold=0.5):
    # 1. Calculate probabilities (use your exact sigmoid function)
    z = X @ w + b
    sig = 1 / (1 + np.exp(-z))

    # 2. Convert to strict 0s and 1s based on threshold
    predictions = (sig >= threshold).astype(int)

    return predictions
```

```python
prediction = predict(x_raw[:700, ],in_w,in_b)
```

```python
def calculate_accuracy(y_true, y_pred):
    # Compare true labels to predicted labels
    accuracy = np.mean(y_true == y_pred) * 100
    return accuracy
```

```python
calculate_accuracy(y[:700, ],prediction)
```
