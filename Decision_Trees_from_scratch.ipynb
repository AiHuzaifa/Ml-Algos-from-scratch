```python
import numpy as np

# Feature 0: Age, Feature 1: Salary
X = np.array([
    [22, 40000],
    [28, 60000],
    [45, 80000],
    [20, 20000],
    [55, 120000],
    [32, 55000],
    [38, 75000],
    [25, 30000],
    [50, 100000],
    [40, 65000]
])

# Target: Bought Car (0 or 1)
y = np.array([0, 0, 1, 0,1, 0, 1, 0, 1, 1])
```

```python
def calculate_gini(y_data):
  if len(y_data) == 0:
        return 0.0
  unique_val , count_val = np.unique(y_data,return_counts=True)
  total_values = np.sum(count_val)
  probabilities = count_val / len(y_data)

    # Gini = 1 - sum(p^2)
  gini = 1.0 - np.sum(probabilities ** 2)
  return gini
```

```python
calculate_gini(y[1:6])
```

```python

def calculate_ig(y_parent,y_left,y_right):
  weighted_left = len(y_left)/len(y_parent)
  weighted_right = len(y_right)/len(y_parent)

  gini_left = calculate_gini(y_left)
  gini_right = calculate_gini(y_right)
  gini_parent = calculate_gini(y_parent)

  information_gain = gini_parent - ((weighted_left*gini_left) + (weighted_right*gini_right))
  return information_gain

```

```python
def get_best_split(X,y):
  best_ig = -1
  best_feature = None
  best_threshold = None

  n_rows, n_cols = X.shape
  for col in range(n_cols):
    thresholds = np.unique(X[:,col])

    for threshold in thresholds:
      print("Threshold",threshold)
      th_left = X[:,col] <= threshold
      th_right = X[:,col] > threshold

      y_th_left = y[th_left]
      y_th_right = y[th_right]
      if len(y_th_left) == 0 or len(y_th_right) == 0:
        continue
      ig = calculate_ig(y,y_th_left,y_th_right)

      if ig>best_ig:
        best_ig = ig
        best_feature = col
        best_threshold = threshold

  return best_feature, best_threshold, best_ig



```

```python
best_col, best_thresh, best_ig = get_best_split(X, y)

print(f"Parent Gini: {calculate_gini(y):.3f}")
print(f"--- WINNING SPLIT ---")
print(f"Best Column to Split: Column {best_col}")
print(f"Best Threshold (Cut): <= {best_thresh}")
print(f"Information Gain:     {best_ig:.3f}")
```
