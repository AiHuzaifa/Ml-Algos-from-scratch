```python
import numpy as np

# Feature 0: Age, Feature 1: Salary
X = np.array([
    [22, 40000],
    [28, 60000],
    [45, 80000],
    [20, 20000],
    [55, 120000],
    [32, 55000],
    [38, 75000],
    [25, 30000],
    [50, 100000],
    [40, 65000]
])

# Target: Bought Car (0 or 1)
y = np.array([0, 0, 1, 0,1, 0, 1, 0, 1, 1])
```

```python
n_samples = len(X)
```

```python
rows, columns = X.shape
```

```python
def find_gini(y_data):
  if len(y_data) == 0:
        return 0.0
  unique_val,val_count = np.unique(y_data , return_counts=True)
  total_val = np.sum(val_count)

  probability = val_count/len(y_data)
  gini = 1 - np.sum((probability**2))
  return gini
```

```python
def find_ig(y_left,y_right,y_parent):
     left_gini = find_gini(y_left)
     right_gini = find_gini(y_right)
     parent_gini = find_gini(y_parent)

     weight_left = len(y_left)/len(y_parent)
     weight_right = len(y_right)/len(y_parent)

     ig = parent_gini - (left_gini*weight_left + right_gini*weight_right)
     return ig

```

```python
best_ig = -1
best_col = None
best_threshold = None

for col in range(columns):
  thresholds = np.unique(X[:,col])
  for threshold in thresholds:
    th_left = X[:,col] <= threshold
    th_right = X[:,col] > threshold

    left_y = y[th_left]
    right_y = y[th_right]

    information_gain = find_ig(left_y,right_y,y)
    if len(left_y) == 0 or len(right_y) == 0:
        continue
    if information_gain > best_ig:
          best_ig = information_gain
          best_col = col
          best_threshold = threshold
          # --- NEW: CAPTURE THE MAJORITY VOTE ---

            # 1. Find the most frequent class in the left box
          left_classes, left_counts = np.unique(left_y, return_counts=True)
            # np.argmax finds the index of the highest count
          prediction_left = left_classes[np.argmax(left_counts)]

            # 2. Find the most frequent class in the right box
          right_classes, right_counts = np.unique(right_y, return_counts=True)
          prediction_right = right_classes[np.argmax(right_counts)]


    # information_gain = find_ig(left_gini,right_gini,)
print(f"Best Information Gain : {best_ig}")
print(f"Best Column : {best_col}")
print(f"Best threshold : {best_threshold}")
```

```python
def predict_new_point(new_data_point):
    # Step 1: Look only at the specific column that won the Greedy Search
    value_to_check = new_data_point[best_col]

    # Step 2: Route the data based on the winning threshold
    if value_to_check <= best_threshold:
        return prediction_left
    else:
        return prediction_right

# --- Let's test it! ---
# Creating a brand new person: Age 30 (Col 0), Salary 45000 (Col 1)
new_person = np.array([35, 45000])

result = predict_new_point(new_person)
print(f"Prediction for new person {new_person}: Class {result}")
```
